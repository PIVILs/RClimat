```python
from glob import glob
from win32com.client import Dispatch

""" Функция xls в xlsx"""

def convert_xls_to_xlsx(oldName:str, newName:str):
    oldName = os.path.abspath(oldName)
    newName = os.path.abspath(newName)
    xlApp = Dispatch('Excel.Application')
    wb = xlApp.Workbooks.Open(oldName)
    wb.SaveAs(newName,51)
    wb.Close(True)

allFiles = [a for a in glob("XML\*.xls")]
for file in allFiles:
    print(file)
    new_path = '.\\XLS\\' + str(file.split('\\')[-1][:-4]) + '.xlsx'
    convert_xls_to_xlsx(file, new_path)
    
#convert_xls_to_xlsx('300729', '300729')
```

```python

def read_all_excel_files(folder_path):
    """
    Читает все файлы Excel в указанной папке.

    Args:
        folder_path (str): Путь к папке с файлами Excel.

    Returns:
        dict: Словарь, где ключами являются имена файлов,
              а значениями - соответствующие DataFrame.
    """
    excel_files = {}
    for filename in os.listdir(folder_path):
        if filename.endswith((".xls", ".xlsx")):
            file_path = os.path.join(folder_path, filename)
            try:
                                
                df = pd.read_excel(file_path)
                excel_files[filename] = df
                filename_new = str(df.columns[2])
                
                allFiles = [a for a in glob("XML\*.xls")]
                for file in allFiles:
                    new_path = '.\\XLS\\' + str(file.split('\\')[-1][:-4]) + '.xlsx'
                convert_xls_to_xlsx(file_path, filename_new)
                
                print(f"Успешно прочитан файл: {filename_new}")
                                
            except Exception as e:
                print(f"Ошибка при чтении файла {filename}: {e}")
            
    return excel_files

# Пример использования
#folder_path = "путь/к/вашей/папке"  # Замените на реальный путь
#data = read_all_excel_files(folder_path)

#  Теперь у вас есть словарь 'data', где ключи - имена файлов,
#  а значения - соответствующие DataFrame.
#  Например, чтобы получить DataFrame из файла "example.xlsx":
#  if "example.xlsx" in data:
#      df = data["example.xlsx"]
#      print(df.head())
```
